Simple statistical imputation


mean imputation

```python
from sklearn.impute import SimpleImputer
mean_imputer=SimpleImputer(strategy="mean")
df_mean=students.copy()
df_mean[["Attendance","Marks"]]=mean_imputer.fit_transform(df_mean[["Attendance","Marks"]])
print(df_mean)
```

```python
from sklearn.impute import SimpleImputer
median_imputer=SimpleImputer(strategy="median")
df_median=students.copy()
df_median[["Attendance","Marks"]]=median_imputer.fit_transform(df_median[["Attendance","Marks"]])
print(df_median)
```

Most -frequent-value impulation

```python
from sklearn.impute import SimpleImputer
mode_imputer=SimpleImputer(missing_values=None,strategy="most_frequent")
df_mode=students.copy()
df_mode[["Name"]]=mode_imputer.fit_transform(df_mode[["Name"]])
print(df_mode)
```

```python
from sklearn.impute import SimpleImputer
const_imputer=SimpleImputer(strategy="constant",fill_value=0)
df_const=students.copy()
df_const[["Marks","Attendance"]]=const_imputer.fit_transform(df_const[["Marks","Attendance"]])
print(df_const)
```

KNN imputation


```python
from sklearn.impute import KNNImputer
knn_imputer=KNNImputer(n_neighbors=2)
df_knn=students.copy()
df_knn[["Marks","Attendance"]]=knn_imputer.fit_transform(df_knn[["Marks","Attendance"]])
print(df_knn)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Marks': [80, 90, np.nan, 70, 100]
})

# Method 1: Drop missing rows
df_drop = df.dropna()

print("After dropping:")
print(df_drop)

# Method 2: Fill missing marks with median
df_fill = df.copy()
df_fill['Marks'] = df_fill['Marks'].fillna(df_fill['Marks'].median())

print("After filling with median:")
print(df_fill)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D'],
    'Marks': [80, np.nan, 90, np.nan]
})

print("Original Data:")
print(df)

# Fill missing values with median
median_marks = df['Marks'].median()
df['Marks'] = df['Marks'].fillna(median_marks)

print("\nCleaned Data:")
print(df)

# Document cleaning decision
print("\nData Cleaning Report:")
print("Missing Marks filled using median:", median_marks)
print("Missing values after cleaning:", df.isnull().sum().sum())
```
