```python
import pandas as pd
import numpy as np
import matplotlib .pyplot as plt
```

```python
# visualization
#graphs
# 1. line graph
# 2. bar graph
# 3. barh
# 4.scatter
# box plot
#histogrram
#piechart
```

```python
# line graph
subjects=['Python','Stastistics','DBMS','Machine Learning','Data visualization']
marks=[85,78,90,82,88]
```

```python
# create the line graph
plt.plot(subjects,marks)
# set the title
plt.title("Student Performance Analysis")

# label the X- axis
plt.xlabel("Subjects")

#label for y axis
plt.ylabel("Marks")

# apply the grid
plt.grid(True)
plt.xticks(rotation=10)


# display the graph
plt.show()
```

```python
plt.plot(subjects,

         marks,
         color="green",
         linestyle="--",
         linewidth=5,
         marker="o",
         label="students",
         markersize=8,
         markerfacecolor="blue",
         markeredgecolor="red",
)


plt.title("student performance analysis")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.xticks(rotation=30)
plt.grid(True)
plt.show()
```

```python
# Bar chart
# create bar chart
plt.bar(
    subjects,
    marks,
    color="grey",
    edgecolor="green",
    width=0.8
)
plt.figure(figsize=(3,2))#width=8 ,height=5
plt.title("Bar Chart")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=20)
plt.legend()
plt.ylim(0,100)
plt.grid(axis="y",alpha=0.5,linestyle="--") # alpha it specifies the transparency
for i,value in enumerate(marks):
  plt.text(i,value+1,str(value),ha="center")
  plt.ylim(0,100)
plt.show()
```

```python
#create  the horizontal bar chart
plt.figure(figsize=(10,6))
plt.barh(subjects,marks,
        color="pink")
plt.title("horizontal bar chart for students marks ")
plt.show()
```

```python
# Scatter Plot # individual Observation
# Number of hours studied by the student
study_hours=[1,2,3,4,5,6,7,8]
marks=[35,40,45,50,55,68,79,90]
# create the scatter plot
plt.scatter(study_hours,
            marks,
            s=150, # size
            c="red" ,# c is color
            marker="o",#type of marker os circle
            alpha=0.8,# this transparency for the marker
            label="Students",
            zorder=3 # displat the marker above the grid
            )
plt.title("Study_hours Vs Examination Marks", fontsize=20)
plt.xlabel("StudHours",fontsize=15,fontweight=10)
plt.ylabel("Examination mark")
plt.xticks(study_hours)
plt.xlim(0,9)
plt.ylim(0,100)
plt.grid(True,linestyle="--",alpha=0.8)
plt.legend()
plt.show()
```
