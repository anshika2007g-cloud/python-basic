```python
def add(*nums):
  total=0
  for x in nums:
    total += x
  return (total)
res1 =add(10, 20)
print(res1)
res2 =add(1,20,30,40,50)
print(res2)
```

```python
import numpy as np
n = int(input("How many numbers"))
numbers = []
for i in range(n):
  x = int(input("Enter a number"))
  numbers.append(x)

arr = np.array(numbers)
print("shape", arr.shape)
print("Dim",arr.ndim)
print("type",type(arr))
print("Array type",arr.dtype)
```

Reshaping


```python
data=np.arange(1,13)
a=data.reshape(3,4)
b=data.reshape(4,3)
c=data.reshape(2,-1)
print(f'1-D data \n{data}')
print(f'2-D data \n{a}')
print(f'2-D data \n{b}')
print(f'2-D data \n{c}')
```

flatten() and ravel

```python
data =np.array([[1,2,3],[4,5,6]])
print("orig data",data.ndim)
flat1 = data.flatten()
print("Flatten flat1",flat1)
flat2 = data.ravel()
print("ravel flat2",flat2)
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.concatenate((a,b))
print("Concatenate \n",c)
print("vstack \n", np.vstack((a,b)))
print("Hstack \n", np.hstack((a,b)))
```
